# Task C — 22.5° Perspective View

## Objective

Generate a view of the port from a slightly right-side camera position, 100 cm from the port centre, at 22.5°.

## Approach

The port is modeled as a planar 40 cm × 40 cm surface. A virtual pinhole camera is positioned 100 cm from the port centre at 22.5°. The projected port corners are calculated and a homography is used to generate the perspective view.
