# Task A — Rotation Estimation

## Objective

Estimate the rotation angle of an arbitrarily rotated satellite-port image using image features.

## Approach

1. Detect ORB keypoints and descriptors.
2. Match features using a BF Hamming matcher.
3. Use RANSAC with `estimateAffinePartial2D()` to reject incorrect matches.
4. Extract the rotation angle from the affine transformation matrix.
